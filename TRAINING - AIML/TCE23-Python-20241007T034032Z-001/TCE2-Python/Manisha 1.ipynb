```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
features = df.drop('diagnosis', axis=1)
scaled_features = scaler.fit_transform(features)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(scaled_features, df['diagnosis'], test_size=0.2, random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Train the Random Forest model
rf = RandomForestClassifier()
rf.fit(X_train, y_train)

# Make predictions
rf_pred = rf.predict(X_test)

# Calculate accuracy
rf_accuracy = accuracy_score(y_test, rf_pred)
print(f'Random Forest Accuracy: {rf_accuracy}')

# Calculate precision
rf_precision = precision_score(y_test, rf_pred, pos_label='M')
print(f'Random Forest Precision: {rf_precision}')

# Calculate recall
rf_recall = recall_score(y_test, rf_pred, pos_label='M')
print(f'Random Forest Recall: {rf_recall}')

# Calculate F1 score
rf_f1 = f1_score(y_test, rf_pred, pos_label='M')
print(f'Random Forest F1 Score: {rf_f1}')

# Generate confusion matrix
rf_cm = confusion_matrix(y_test, rf_pred, labels=['B', 'M'])
print(f'Confusion Matrix:\n{rf_cm}')
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Train the SVM model
svm = SVC()
svm.fit(X_train, y_train)

# Make predictions
svm_pred = svm.predict(X_test)

# Calculate accuracy
svm_accuracy = accuracy_score(y_test, svm_pred)
print(f'SVM Accuracy: {svm_accuracy}')

# Calculate precision
svm_precision = precision_score(y_test, svm_pred, pos_label='M')
print(f'SVM Precision: {svm_precision}')

# Calculate recall
svm_recall = recall_score(y_test, svm_pred, pos_label='M')
print(f'SVM Recall: {svm_recall}')

# Calculate F1 score
svm_f1 = f1_score(y_test, svm_pred, pos_label='M')
print(f'SVM F1 Score: {svm_f1}')

# Generate confusion matrix
svm_cm = confusion_matrix(y_test, svm_pred, labels=['B', 'M'])
print(f'Confusion Matrix:\n{svm_cm}')
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Train the KNN model
knn = KNeighborsClassifier()
knn.fit(X_train, y_train)

# Make predictions
knn_pred = knn.predict(X_test)

# Calculate accuracy
knn_accuracy = accuracy_score(y_test, knn_pred)
print(f'KNN Accuracy: {knn_accuracy}')

# Calculate precision
knn_precision = precision_score(y_test, knn_pred, pos_label='M')
print(f'KNN Precision: {knn_precision}')

# Calculate recall
knn_recall = recall_score(y_test, knn_pred, pos_label='M')
print(f'KNN Recall: {knn_recall}')

# Calculate F1 score
knn_f1 = f1_score(y_test, knn_pred, pos_label='M')
print(f'KNN F1 Score: {knn_f1}')

# Generate confusion matrix
knn_cm = confusion_matrix(y_test, knn_pred, labels=['B', 'M'])
print(f'Confusion Matrix:\n{knn_cm}')
```

```python
models = ['Random Forest', 'SVM', 'KNN']
accuracies = [rf_accuracy, svm_accuracy, knn_accuracy]
precisions = [rf_precision, svm_precision, knn_precision]
recalls = [rf_recall, svm_recall, knn_recall]
f1_scores = [rf_f1, svm_f1, knn_f1]
confusion_matrices = [rf_cm, svm_cm, knn_cm]

for model, accuracy, precision, recall, f1, cm in zip(models, accuracies, precisions, recalls, f1_scores, confusion_matrices):
    print(f'{model} - Accuracy: {accuracy}, Precision: {precision}, Recall: {recall}, F1 Score: {f1}')
    print(f'Confusion Matrix:\n{cm}\n')

# Determine the best model
best_model_index = f1_scores.index(max(f1_scores))
best_model = models[best_model_index]
print(f'The best model is {best_model} with an F1 Score of {f1_scores[best_model_index]}')
```
