## Lambda Functions
A lambda function is a small **anonymous** function. A lambda function can take **any number of arguments**, but can only have **one expression**

```python
def add(x , y):
    return x+y
print(add(10, 20))
```

```python
mylambdafn = lambda x, y, z: (x+y)*z
print(mylambdafn(10, 20, 2))
print(mylambdafn(3,6,5))
```

### Use Case -  Quadratic Equation

image.png

image.png

If you get a negative value inside the square root, then there will be no real number solution, and therefore no x-intercepts

```python
import math
def quadraticEquation(a, b, c):
    discriminant = math.sqrt(b**2 - 4*a*c)
    x1 = (-1*b + discriminant)/2*a
    x2 = (-1*b - discriminant)/2*a
    return x1, x2
```

```python
x1, x2 = quadraticEquation(2, 3, 1)
print(x1,x2)
```

### Use Case - Check the validity of Email Address


| Character | Description | Example
|----|---|----|
| [ ]  | A set of characters | "a-z"
|  \ | Signals a special sequence. Also used to escape special characters | "\n"
| .  | Any character (except newline character) | "he..o"
| ^  | Starts with | "Dr."
| $  | Ends with | "Singh"
| *  | 0 or more occurance | "bix*"
| +  | 1 or more occurance | "bix+"
| { }  |Exactly the specified number of occurances | "bix{1}"
| \| |Either or | "he\|she"
| \w | Matches any word character | ""
| ( ) |Capture Group

```python
import re
def check(email):
    regex = '^[a-z0-9]+[\._]?[a-z0-9]+[@]\w+[.]\w{2,3}$'
    if(re.search(regex,email)):
        print(email, " : Valid Email")
    else:
        print(email, " : Invalid Email")
```

```python
email = "content@edureka.co"
check(email)
```

```python
email = "content.team@gmail.com"
check(email)
```

```python
email = "edureka.co"
check(email)
```

## Class and Objects in Python

```python
class Emp():
    '''
    Class: Emp
    Properties: id and name
    Methods: greet
    Constructor: id and name passed as parameters
    '''
    def __init__(self, id, name):
        self.id = id
        self.name = name
    def greet(self, company):
        print('Hi ', self.name, ", Welcome to " + company)
```

```python
empVirat = Emp(1, 'Virat')
empRohit = Emp(2, 'Rohit')

empVirat.greet("BCCI")
empRohit.greet("ABC")
```

```python
class MyClass():
    '''Demonstrates Public and Private methods'''
    def myPublicMethod(self):
        print('Public Method')
    def __myPrivateMethod(self):
        print('Private Method')
    def myPublicMethod2(self):
        print("calling the private method from myPublicMethod2")
        self.__myPrivateMethod()

obj = MyClass()
obj.myPublicMethod()
obj.myPublicMethod2()
```

## Handling Exceptions

```python
try:
    a = 10
    b = int(input('enter the number:'))
    print(a, b, a/b)
except Exception as e:
    print("Error:",e)
finally:
    print("All cleanup statements here")
```

```python
try:
    a = 10
    b = int(input('enter the number:'))
    # if b == 0:
    #     raise ZeroDivisionError
    print(a, b, a/b)
except ZeroDivisionError:
    print("Zero div error: Please enter a non zero number")
except ValueError:
    print("Value Error: Please enter a number only")
except Exception as e:
    print("Error:",e)
finally:
    print("All cleanup statements here")
```

```python
# define Python user-defined exceptions
# Base class for other exceptions
class Error(Exception):
    pass


# Raised when the input value is too small
class ValueSmallError(Error):
    pass


# Raised when the input value is too large
class ValueLargeError(Error):
    pass


number = 10
while True:
    try:
        in_num = int(input("Enter a number: "))
        if in_num < number:
            raise ValueSmallError
        elif in_num > number:
            raise ValueLargeError
            #break
        elif in_num == number:
            break
    except ValueSmallError:
        print("This value is too small, try again!")
        print()
    except ValueLargeError:
        print("This value is too large, try again!", end ="\n")

print("Congratulations! You guessed it correctly.")
```
