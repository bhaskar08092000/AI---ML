#### Statistical Functions in NumPy

```python
#amin() & amax() - Returns the minimum & the maximum from the elements in the given array along the given axis
# Minimum of array, if axis is None; the result is an array of dimension a.ndim – 1, if axis is mentioned

arr = np.array([[5,4,6],[1,3,9],[7,2,8]])
print(arr)
```

```python
print(np.amin(arr))
print(np.amin(arr, 0))  # minimum of each column
print(np.amin(arr, 1))  # minimum of each row
```

```python
print(np.amax(arr))
print(np.amax(arr, 0))  # max of each column
print(np.amax(arr, 1))  # max of each row
```

```python
arr = np.array([[4,9,2],
              [3,1,2]])
arr
```

```python
print('mean:',arr.mean())
print('Std dev:',arr.std()) #std() - Returns the Standard Deviation of the given data. It is the square-root of variance
print('min value',arr.min())
print('max value', arr.max())

print('median', np.median(arr))
```

```python
#mean() - Returns the Mean of the given data

arr = np.array([[0,5,7],
              [4,9,1],
              [3,1,7]])
print(np.mean(arr))

print(np.mean(arr,0)) #Mean along axis - 0 (mean of each column)

print(np.mean(arr,1)) #Mean along axis - 1 (mean of each row)
```

### Task 4:Create average user rating matrix using movies_matrix
The average_user_rating_matrix will have 4 columns:
* ‘Movie ID’
* ‘Average rating’
* ‘Number of ratings’
* ‘Standard deviation of ratings’

```python
# picking up one column which will be the rating of that movie by 110 users
i = 0
m = movie_matrix[:,i]
print(m)
m = m[m>=0]
print(m)
```

```python
final_movie_rating = []
total_num_rating = m.size
rating = m.mean()
standard_deviation = m.std()
final_movie_rating.append([new_movies_id[i],rating,total_num_rating,standard_deviation])
final_movie_rating
```

```python
# putting the above together in a a loop
final_movie_rating = []
for i in range(1050):
    #each movie's rating
    m = movie_matrix[:,i] # basically picking up one column at a time, which will be rating of that movie by all 110 users
    m = m[m>=0]
    #total_rate = m.sum()
    total_num_rating = m.size
    rating = m.mean()
    standard_deviation = m.std()
    final_movie_rating.append([new_movies_id[i],rating,total_num_rating,standard_deviation])
```

### Task 5: Convert the final average movie ratings to range from 0 to 10
The minimum rating converts to 0 and maximum to 10 , and the other values in between

```python
#Converting the average of movie ratings to have the range 0 to 10
x = final_movie_rating[:,1]
old_range = x.max()-x.min()
new_range = (10 - 0)
final_movie_rating[:,1] = ((x - x.min())*(new_range/old_range) + 0)
```

#### numpy.argsort
* Using numpy.argsort() in descending order returns the indicies of the highest n values in a numpy.array.
* USE LIST SLICING TO USE NUMPY ARGSORT IN DESCENDING ORDER
* Use numpy.argsort(an_array) to sort the indices of an_array in ascending order. Use the syntax ranked[::-1] with ranked as the previous result to reverse ranked

```python
arr = np.array([ 31, 5,  8, 21, 15, 18, 9])
print ("Original unsorted array: ", arr)
arr_index_sorted = np.argsort(arr)
print ("Sorted indices : ", arr_index_sorted)
print("Sorted array : ", arr[arr_index_sorted])
```

```python
arr = np.array([ 31, 5,  8, 21, 15, 18, 9])
print ("Original unsorted array: ", arr)
arr_index_sorted = arr.argsort()
print ("Sorted indices ascending: ", arr_index_sorted)
arr_index_sorted = arr.argsort()[::-1]
print ("Sorted indices descending : ", arr_index_sorted)
print("Sorted array descending : ", arr[arr_index_sorted])
```
