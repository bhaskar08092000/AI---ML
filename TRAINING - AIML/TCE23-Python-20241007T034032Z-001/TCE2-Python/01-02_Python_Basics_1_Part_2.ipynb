## Check if the number is prime  (homework)

```python
#Check if a number is prime
num = int(input("Enter a number: "))
is_prime = True
for i in range(2, num):
  print("Iteration:", i)
  if num % i == 0:
    is_prime = False  # number is not prime
    break

if is_prime:
  print(num, "is prime")
else:
  print(num, "is not a prime")
```

## Find the cube of a number (homework)

```python
# Find the cube of a number
while True:
  value = input("\nEnter a number: ")
  if value =='q':
    # if input is 'q' exit from the while loop
    print("Exiting program (break statement executed)...")
    break

  if not value.isdigit():
    # if input is not a digit move on to the next iteration
    print("Enter digits only (continue statement executed)")
    continue
  value = int(value)
  print("Cube of ", value, "is", value **3) # everything is fine, just print the cube
```

## Functions

```python
def celsius_to_fahr(temp):
  return ((9/5) * temp) + 32
```

```python
celsius_to_fahr(34)
```

```python
celsius_to_fahr(38)
```

```python
def print_msg(count, msg):
  while count > 0:
    print(msg)
    count -= 1
```

```python
print_msg(5, "Today is the 4th Session for Python...please do your homework during the week")
```

```python
def factorial(n):
  f = 1
  if int(n) >= 1:
    for i in range(2, int(n)+1):
      f = f * i
  return f
```

```python
n = input("enter the number:")
factorial(n)
```

### Types of function arguments - keyword, default, variable length arguments

#### Order of arguments

```python
def substring1(mystring, start_index, end_index):
    print("Substring is:", mystring[start_index : end_index])
```

```python
substring1("This is my sample string", 10, 20)
```

```python
substring1(10, 20, "This is my sample string")
```

#### Keyword arguments

```python
#substring1(10, 20, "This is my sample string")
# using keywords
substring1(start_index=10, end_index=20, mystring="This is my sample string")
```

```python
substring1(end_index = 22, start_index=10, mystring="This is my sample string")
```

```python
substring1( "This is my sample string", 10, 22)
```

#### Default arguments

```python
# default
def substring2(mystring, start_index=10, end_index=20):
    print("Substring is:", mystring[start_index : end_index])
```

```python
substring2("This is my sample string", 10, 20)
```

```python
substring2("This is my sample string", 12, 22)
```

```python
substring2("This is my sample string")
```

#### Variable length arguments

```python
def purchaseinfo(product1, *products):
    print("Products purchased:")
    print(products)
    print(product1)
    for item in products:
        print(item)
```

```python
purchaseinfo('laptop', 'desktop')
```

```python
purchaseinfo('laptop', 'desktop', 'tablet', 'mouse', 'iphone')
```
