## List Comprehension

```python
nums=[1,2,3,4,5]
squares=[n*n for n in nums]
print(squares)
```

## Lambda, map, filter, reduce

```python
from functools import reduce
print(list(map(lambda x:x*2,[1,2,3])))
print(list(filter(lambda x:x%2==0,[1,2,3,4])))
print(reduce(lambda a,b:a+b,[1,2,3,4]))
```

## Generator

```python
def gen():
    for i in range(5):
        yield i
for x in gen():
    print(x)
```

## First Non-Repeating Number

```python
from collections import Counter
nums=[1,2,2,3,1,4,5]
f=Counter(nums)
print([n for n in nums if f[n]==1])
```

## Two Sum

```python
nums=[2,7,11,15]
target=9
seen={}
for i,n in enumerate(nums):
    if target-n in seen:
        print(seen[target-n],i)
        break
    seen[n]=i
```

## Running Average

```python
nums=[10,20,30,40]
total=0
for i,n in enumerate(nums,1):
    total+=n
    print(total/i)
```
